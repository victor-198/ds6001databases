### Answers for Problem 3 Part a

#### i) `charworks` and `works`
1. Join on `workid`.
2. Each row in `charworks` references one work, while one work appears in many `charworks` rows (many characters per work).
3. Relationship: **many-to-one** from `charworks` to `works` (equivalently one-to-many from `works` to `charworks`).
4. DBML: `Ref: charworks.workid > works.workid`

#### ii) `characters` and `charworks`
1. Join on `charid`.
2. Each `charworks` row references one character, while one character may appear in many works.
3. Relationship: **one-to-many** from `characters` to `charworks`.
4. DBML: `Ref: characters.charid < charworks.charid`

#### iii) `works` and `paragraphs`
1. Join on `workid`.
2. One work has many lines/paragraphs; each paragraph belongs to one work.
3. Relationship: **one-to-many** from `works` to `paragraphs`.
4. DBML: `Ref: works.workid < paragraphs.workid`

#### iv) `chapters` and `works`
1. Join on `workid`.
2. One work has many chapters/scenes; each chapter belongs to one work.
3. Relationship: **many-to-one** from `chapters` to `works` (or one-to-many from `works` to `chapters`).
4. DBML: `Ref: chapters.workid > works.workid`

#### v) `paragraphs` and `chapters`
1. Join on `chapterid`.
2. One chapter contains many paragraphs; each paragraph belongs to one chapter.
3. Relationship: **many-to-one** from `paragraphs` to `chapters`.
4. DBML: `Ref: paragraphs.chapterid > chapters.chapterid`

#### vi) `characters` and `paragraphs`
1. Join on `charid`.
2. One character can speak many lines; each paragraph is spoken by one character (including group-speaker IDs like Chorus/Witches treated as one character).
3. Relationship: **one-to-many** from `characters` to `paragraphs`.
4. DBML: `Ref: characters.charid < paragraphs.charid`

````python
def pandas_df_to_dbml(df: pd.DataFrame, table_name: str) -> str:
    """
    Converts a pandas DataFrame to a DBML string.

    Args:
        df: The pandas DataFrame to convert.
        table_name: The name of the table in the DBML schema.

    Returns:
        A DBML string representing the DataFrame schema.
    """

    dbml_string = f"Table {table_name} {{\n"

    for column_name, column_type in df.dtypes.items():
        dbml_type = map_pandas_dtype_to_dbml_type(column_type)
        dbml_string += f"  {column_name} {dbml_type}\n"

    dbml_string += "}\n"
    return dbml_string

def map_pandas_dtype_to_dbml_type(dtype) -> str:
    """Maps a pandas dtype to a DBML type."""
    dtype_name = str(dtype)
    if "int" in dtype_name:
      return "int"
    if "float" in dtype_name:
      return "float"
    if "datetime" in dtype_name:
        return "datetime"
    return "varchar"
````

### Part a
Initialize a new Shakespeare database using SQlite via the `sqlite3` package. Next use the `pd.to_sql()` method to add the five Shakespeare dataframes to this database. Then, to prove that this worked, issue the following SQL query to the database which should display a dataframe listing all characters from Shakespeare's plays with more than 200 lines of dialogue (and the Poet too):
```
SELECT charname, description, speechcount
FROM characters
WHERE speechcount > 200
```
Finally, after running the query, use the `.commit()` and `.close()` methods on the Python variable containing the database connection to save your changes and prevent further changes.

[8 points]

````python
repo = 'https://github.com/jkropko/DS-6001/raw/master/localdata/'
works = pd.read_csv(repo + 'Works.csv')
characters = pd.read_csv(repo + 'Characters.csv')
chapters = pd.read_csv(repo + 'Chapters.csv')
paragraphs = pd.read_csv(repo + 'Paragraphs.csv')

# convert column names to lowercase (needed for postgreSQL to work properly)
characters.columns = characters.columns.str.lower() 
chapters.columns = chapters.columns.str.lower()
paragraphs.columns = paragraphs.columns.str.lower()
works.columns = works.columns.str.lower()

# works in the characters tables is a comma separated list. 
# Break it out into multiple rows in a new table
charworks = characters[['charid', 'works']]
charworks['works'] = charworks['works'].str.split(',')
charworks = charworks.explode('works')
charworks = charworks.rename({'works':'workid'}, axis=1)
characters = characters.drop('works', axis=1)

#Remove empty rows
chapters = chapters.query("~chapterid.isnull()")
paragraphs = paragraphs.query("~paragraphid.isnull()")
charworks = charworks.query("~workid.isnull()")

# Add chapterid to paragraphs
paragraphs = pd.merge(paragraphs, 
                      chapters.drop('description', axis=1),
                      how='inner', 
                      on=['workid', 'section', 'chapter'])

#Remove unnecessary columns
paragraphs = paragraphs.drop(['paragraphtype', 'section', 'chapter'], 
                             axis=1) 
````

````python
shakespeare_db = sqlite3.connect("shakespeare.db") 

characters.to_sql('characters', shakespeare_db, index=False, chunksize=1000, if_exists='replace')
works.to_sql('works', shakespeare_db, index=False, chunksize=1000, if_exists='replace')
paragraphs.to_sql('paragraphs', shakespeare_db, index=False, chunksize=1000, if_exists='replace')
charworks.to_sql('charworks', shakespeare_db, index=False, chunksize=1000, if_exists='replace')
chapters.to_sql('chapters', shakespeare_db, index=False, chunksize=1000, if_exists='replace')
````

````python
myquery = '''
SELECT charname, description, speechcount
FROM characters
WHERE speechcount > 200
'''

characters_200 = pd.read_sql_query(myquery, shakespeare_db)
display(characters_200)

shakespeare_db.commit()
shakespeare_db.close()
````
